```python
import pandas as pd

df = pd.read_csv("kegiatan_2025.csv")

# Ambil kode pusat riset dari ID Riset
df["Pusat Riset"] = df["ID Riset"].str.split("_").str[0]

# Jumlah kegiatan per pusat riset
print(df.groupby("Pusat Riset").size())

# Jumlah kegiatan per kelompok riset
print(df.groupby(["Pusat Riset", "Kelompok Riset"]).size())
```

Memeriksa Dataset

```python
# Melihat ukuran dataset
print("Ukuran dataset:", df.shape)

# Melihat nama kolom
print("\nNama kolom:")
print(df.columns.tolist())

# Melihat 5 baris pertama
display(df.head())

# Memeriksa tipe data dan jumlah data yang terisi
print("\nInformasi dataset:")
df.info()

# Memeriksa data kosong
print("\nJumlah data kosong per kolom:")
print(df.isnull().sum())
```

Memeriksa kelengkapan deskripsi kegiatan

```python
# Menghitung deskripsi yang kosong atau hanya berisi spasi
deskripsi = df["Kegiatan Riset"].fillna("").astype(str).str.strip()

print("Total kegiatan:", len(df))
print("Deskripsi terisi:", (deskripsi != "").sum())
print("Deskripsi kosong:", (deskripsi == "").sum())

# Melihat contoh deskripsi kegiatan
display(
    df.loc[deskripsi != "", ["ID Riset", "Kelompok Riset", "Kegiatan Riset"]]
      .head(10)
)
```

Membersihkan data

```python
# Menyalin data agar dataset asli tetap utuh
df_clean = df.copy()

# Membersihkan teks
df_clean["Kegiatan Riset"] = (
    df_clean["Kegiatan Riset"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# Mengambil data dengan deskripsi yang terisi
df_text = df_clean[df_clean["Kegiatan Riset"] != ""].copy()

# Menghapus duplikasi deskripsi untuk pemeriksaan awal
print("Jumlah kegiatan dengan deskripsi:", len(df_text))
print(
    "Jumlah deskripsi unik:",
    df_text["Kegiatan Riset"].nunique()
)

display(df_text.head())
```

Membuat representasi TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Mengubah deskripsi kegiatan menjadi vektor TF-IDF
tfidf = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2)
)

X = tfidf.fit_transform(df_text["Kegiatan Riset"])

print("Ukuran matriks TF-IDF:", X.shape)
print("Jumlah kegiatan:", X.shape[0])
print("Jumlah fitur teks:", X.shape[1])
```

Menghitung cosine similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Menghitung kemiripan setiap pasangan kegiatan
similarity_matrix = cosine_similarity(X)

# Menampilkan matriks kemiripan
similarity_df = pd.DataFrame(
    similarity_matrix,
    index=df_text["ID Riset"],
    columns=df_text["ID Riset"]
)

display(similarity_df.round(2))
```

Menampilkan pasangan kegiatan yang paling mirip

```python
import numpy as np

similarity_values = []

for i in range(len(df_text)):
    for j in range(i + 1, len(df_text)):
        similarity_values.append({
            "ID Kegiatan 1": df_text.iloc[i]["ID Riset"],
            "Kegiatan 1": df_text.iloc[i]["Kegiatan Riset"],
            "ID Kegiatan 2": df_text.iloc[j]["ID Riset"],
            "Kegiatan 2": df_text.iloc[j]["Kegiatan Riset"],
            "Kemiripan": similarity_matrix[i, j]
        })

pasangan_mirip = pd.DataFrame(similarity_values)

if not pasangan_mirip.empty:
    pasangan_mirip = pasangan_mirip.sort_values(
        "Kemiripan", ascending=False
    )

    display(pasangan_mirip.head(10).round({"Kemiripan": 3}))
else:
    print("Belum ada pasangan kegiatan yang dapat dibandingkan.")
```

Percobaan awal clustering menggunakan K-Means

K-Means mengelompokkan kegiatan berdasarkan kedekatan vektor TF-IDF. Kita akan mencoba beberapa jumlah cluster, lalu menghitung silhouette score jika memungkinkan.

Mencoba beberapa jumlah cluster

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

n_data = X.shape[0]
hasil_evaluasi = []
model_terbaik = None
label_terbaik = None
k_terbaik = None
skor_terbaik = -1

if n_data < 3:
    print(
        "Data yang memiliki deskripsi kurang dari 3 kegiatan. "
        "Tambahkan data sebelum melakukan evaluasi clustering."
    )

elif X.shape[1] == 0:
    print("Tidak ada fitur teks yang bisa digunakan untuk clustering.")

else:
    # Mencoba jumlah cluster dari 2 sampai maksimal 5
    batas_k = min(5, n_data - 1)

    for k in range(2, batas_k + 1):
        model = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        labels = model.fit_predict(X)

        # Silhouette score tidak dapat dihitung jika hanya ada 1 cluster
        # atau setiap data menjadi cluster tersendiri
        if len(set(labels)) < 2 or len(set(labels)) >= n_data:
            continue

        skor = silhouette_score(
            X,
            labels,
            metric="cosine"
        )

        hasil_evaluasi.append({
            "Jumlah Cluster": k,
            "Silhouette Score": skor
        })

        print(f"k = {k}, silhouette score = {skor:.4f}")

        if skor > skor_terbaik:
            skor_terbaik = skor
            model_terbaik = model
            label_terbaik = labels
            k_terbaik = k

    if hasil_evaluasi:
        print("\nRingkasan evaluasi:")
        display(
            pd.DataFrame(hasil_evaluasi)
            .sort_values("Silhouette Score", ascending=False)
            .reset_index(drop=True)
        )

        print("Jumlah cluster dengan skor tertinggi:", k_terbaik)
    else:
        print(
            "Belum ada konfigurasi yang dapat dievaluasi. "
            "Periksa kembali jumlah dan variasi data."
        )
```

Silhouette score memiliki rentang −1 sampai 1.
- Nilai yang mendekati 1 menunjukkan bahwa data cenderung lebih dekat dengan cluster-nya sendiri dibandingkan dengan cluster lain.
- Nilai sekitar 0 menunjukkan bahwa beberapa data berada di antara cluster.
- Nilai negatif dapat menunjukkan bahwa beberapa data mungkin lebih dekat dengan cluster lain.

Menampilkan hasil clustering

```python
if label_terbaik is not None:
    hasil_clustering = df_text[
        ["ID Riset", "Pusat Riset", "Kelompok Riset", "Kegiatan Riset"]
    ].copy()

    hasil_clustering["Cluster"] = label_terbaik

    # Menghitung jumlah kegiatan dalam setiap cluster
    print("Jumlah kegiatan per cluster:")
    display(
        hasil_clustering["Cluster"]
        .value_counts()
        .sort_index()
        .rename_axis("Cluster")
        .reset_index(name="Jumlah Kegiatan")
    )

    # Melihat isi setiap cluster
    for cluster_id in sorted(hasil_clustering["Cluster"].unique()):
        print(f"\n=== Cluster {cluster_id} ===")

        display(
            hasil_clustering[
                hasil_clustering["Cluster"] == cluster_id
            ][
                ["ID Riset", "Kelompok Riset", "Kegiatan Riset"]
            ]
        )
else:
    print(
        "Belum ada hasil clustering yang bisa ditampilkan. "
        "Periksa hasil Cell 8 terlebih dahulu."
    )
```

Membandingkan hasil dengan kelompok riset asli
Dataset kamu juga memiliki kolom Kelompok Riset. Informasi ini bisa digunakan untuk memeriksa apakah cluster yang terbentuk memiliki pola yang menarik dibandingkan dengan kelompok riset yang sudah ditentukan oleh BRIN.

Membuat tabel perbandingan

```python
if label_terbaik is not None:
    tabel_perbandingan = pd.crosstab(
        hasil_clustering["Kelompok Riset"],
        hasil_clustering["Cluster"]
    )

    display(tabel_perbandingan)
```
